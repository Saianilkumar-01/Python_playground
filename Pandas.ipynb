Series in Pandas

```python
a = [1,2,3,4,5,6]

x = pd.Series(a)
# x = pd.series([1,3,7,4,5,6,77])

print(a,type(a))
print(x,type(x))
```

```python
#print(x[0]) #label

x = pd.Series(a,index=[1,2,3,4,5,6])#create or manage label(index in python for list)
print(x)
print(x[1])
```

```python
y = {
    "day-01" : 450,
    "day-02" : 488,
    "day-03" : 677
}
diet = pd.Series(y)
print(diet)
```
