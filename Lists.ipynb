```python
""" Lists are the built-in data structres used in python to stroe an ordered collection of items.
    > Lists are mutable(List can be modified after creation).
    > Lists stores the input in an ordered manner.
    > Lists are indexed based ehich starts from 0.
    > Lists can store any type of data including other lists and characters.
"""

```

```python
# Creation of Lists:-

a = [1, 2, 3, 4, 5, 10] # Basic

b = [ 2 , "python" , 9.0 , [1 , 4 , 3]] # different data types

c = list("Python") # using list() function

d = [0] * 5 # using multiplication operator
```

```python
# Adding elements to a list:-

# append(value) -- method used to insert an element at the end of the  list.
# a = [1 , 2 ,3 , 4 , 5 , 10]
a.append(6) # a = [1 , 2 ,3 , 4 , 5 , 10 ,  6]

# insert(index, value) -- method used to insert an element at a specific index in the list.
a.insert(2, 10) # a = [1 , 2 ,10 , 3 , 4 , 5 ,10 , 6]

# extend(values) -- method used to add multiple elements to the end of the list.
a.extend([7, 8, 9]) # a = [1 , 2 ,10 , 3 , 4 , 5 ,10 , 6 , 7, 8, 9]
```

```python
# Removing elements from a list:-

# remove(value) -- method used to remove the first occurence of the value in list.
a.remove(10) # a = [1 , 2 , 3 , 4 , 5 , 10 , 6 , 7, 8, 9]

# pop(index) -- method used to remove an element at a specific index in the list.
a.pop(2) # a = [1 , 2 , 3 , 4 , 5 , 10 , 6 , 7, 8, 9]

# del(index) -- method used to delete an element at a specific index in list.
del a[2] # a = [1 , 2 , 4 , 5 , 10 , 6 , 7, 8, 9]

# list.clear() -- method used to remove all elements from the list.
a.clear() # a = []
```
