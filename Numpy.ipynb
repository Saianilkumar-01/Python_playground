```python
import numpy as np
```

```python
a = np.array([1,3,4,5,2,234])  #onr dimension array
print(a)
b = np.array([[1,2,3,4,9],
              [5,6,7,8,10]])
print(b)
```

```python
#Shape , size , dimension
print(b.ndim) #dimension
print(b.shape) #shape == rows , coloums
print(b.size) #size ===> no of elements
print(b.itemsize) #itemsize shows the memory allocated or used by the element in array

```

```python
print(b[1,:]) # full row
print(b[:,3]) #full coloumn
print(b[1, 1:-1:2]) # [rowidx , startidx:endidx:stepcount]
```
